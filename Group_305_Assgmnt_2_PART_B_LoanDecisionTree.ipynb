# Loan Sanction Prediction Using Decision Tree and Prolog

## Overview

This notebook demonstrates the process of building a **Decision Tree Classifier** to predict loan sanction status based on attributes such as sex, age, loan term, number of accounts, loan type, and loan area.  

The decision tree rules are extracted and translated into a **Prolog knowledge base** for logical reasoning and classification. The goal is to classify whether a loan application will be sanctioned or not.

---

## Dataset

The loan dataset contains the following attributes:

- `sex`
- `age`
- `Loan Term in Years`
- `Number_of_Accounts`
- `Loan Type`
- `Loan Area`
- `Loan Sanctioned` (target variable: 1 for sanctioned, 0 for not sanctioned)

---


```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, export_text

# Load dataset (Google sheet exported as CSV or Excel)
url = "https://docs.google.com/spreadsheets/d/1aWBRjmEbZrtfJk1whJNK_GKCSTNvwhDSvCTbxc6C5x8/export?format=csv"
df = pd.read_csv(url)

# Check column names
print(df.columns.tolist())

# Define features and target
features = ['sex', 'age', 'Loan Term in Years', 'Number_of_Accounts', 'Loan Type', 'Loan Area']
target = 'Loan Sanctioned'

X = df[features]
y = df[target]

# Convert categorical values into numbers
X_encoded = pd.get_dummies(X)

# Train Decision Tree
clf = DecisionTreeClassifier(criterion="entropy", max_depth=3, random_state=42)
clf.fit(X_encoded, y)

# Display rules
tree_rules = export_text(clf, feature_names=list(X_encoded.columns))
print(tree_rules)
```
